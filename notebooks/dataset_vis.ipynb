```python
def pretty_print_sample(dataset: DatasetBase, x: Tensor, verbose: bool = True) -> str:
    steps = [
        dataset.to_string(x[i]).split("\n")
        for i in range(x.shape[0])
    ]

    x_str = "\n".join(
        "   ".join(
            steps[i][j]
            for i in range(len(steps))
        )
        for j in range(len(steps[0]))
    )

    if verbose:
        print(x_str)

    return x_str
```
